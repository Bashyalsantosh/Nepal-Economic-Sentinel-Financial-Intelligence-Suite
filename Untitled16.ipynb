```python
import pandas as pd
import numpy as np
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
# उदाहरणका लागि सानो डेटासेट
data = {
    'text': [
        "काठमाडौंमा भोलि सार्वजनिक बिदा दिने सरकारको निर्णय",
        "यो समाचार पढेर तपाईलाई अचम्म लाग्नेछ, पैसा डबल हुने तरिका",
        "नेपाल र भारतबीच नयाँ सम्झौतामा हस्ताक्षर",
        "खुसीको खबर! अब सबैले सित्तैमा आइफोन पाउने"
    ],
    'label': ['Real', 'Fake', 'Real', 'Fake']
}

df = pd.DataFrame(data)
```

```python
def clean_nepali_text(text):
    # विराम चिन्ह र विशेष क्यारेक्टर हटाउने
    text = re.sub(r'[^\u0900-\u097F\s]', '', text)
    return text

df['text'] = df['text'].apply(clean_nepali_text)
```

```python
# डेटा बाँडफाँड (Split)
x_train, x_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=7)

# TF-IDF Vectorizer (शब्दलाई अंकमा बदल्ने)
tfidf_vectorizer = TfidfVectorizer(max_df=0.7)
tfidf_train = tfidf_vectorizer.fit_set(x_train)
tfidf_test = tfidf_vectorizer.transform(x_test)

# PassiveAggressiveClassifier प्रयोग (फेक न्यूजका लागि उपयुक्त)
pac = PassiveAggressiveClassifier(max_iter=50)
pac.fit(tfidf_train, y_train)

# नतिजा जाँच गर्ने
y_pred = pac.predict(tfidf_test)
score = accuracy_score(y_test, y_pred)
print(f'Accuracy: {round(score*100,2)}%')
```

```python
import pandas as pd
from newspaper import Article
import time

# तपाईंले भनेका स्रोतहरूको लिस्ट (URLs)
news_links = [
    "https://www.onlinekhabar.com/2024/05/133456",
    "https://setopati.com/politics/324567",
    "https://ratopati.com/story/425678",
    "https://ekantipur.com/news/2024/05/10/56789.html",
    "https://www.bbc.com/nepali/articles/c1234567890"
]

def scrape_nepali_news(url_list):
    compiled_data = []

    for url in url_list:
        try:
            print(f"Scraping: {url}")
            article = Article(url, language='ne')
            article.download()
            article.parse()

            # डेटा स्टोर गर्ने
            compiled_data.append({
                'Source': url.split('//')[-1].split('/')[0], # साइटको नाम निकाल्ने
                'Title': article.title,
                'Text': article.text,
                'Publish_Date': article.publish_date,
                'Label': 'Real' # यी सबै आधिकारिक स्रोत हुन्
            })

            # सर्भरलाई ओभरलोड हुन नदिन थोरै समय पर्खने
            time.sleep(1)

        except Exception as e:
            print(f"Could not scrape {url}: {e}")

    return pd.DataFrame(compiled_data)

# स्क्र्यापिङ सुरु गर्ने
df = scrape_nepali_news(news_links)

# नतिजा हेर्ने
print("\n--- Scraped Data Summary ---")
print(df.head())

# CSV मा सेभ गर्ने (फेक न्यूज मोडेल ट्रेनिङका लागि)
df.to_csv('nepali_news_dataset.csv', index=False)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def scrape_nepali_portal(url):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }

    try:
        response = requests.get(url, headers=headers, timeout=10)
        if response.status_code != 200:
            return None

        soup = BeautifulSoup(response.content, 'html.parser')

        # शीर्षक (Title) निकाल्ने
        title = soup.find('h1').text.strip() if soup.find('h1') else "No Title"

        # समाचारको मुख्य अंश (Text) निकाल्ने
        # धेरैजसो नेपाली पोर्टलमा समाचार <p> ट्याग भित्र हुन्छ
        paragraphs = soup.find_all('p')
        content = " ".join([p.text.strip() for p in paragraphs if len(p.text.strip()) > 20])

        return {
            "URL": url,
            "Title": title,
            "Text": content,
            "Label": "Real"
        }
    except Exception as e:
        print(f"Error scraping {url}: {e}")
        return None

# अहिलेका ताजा समाचारका लिङ्कहरू (यी लिङ्कहरू काम गर्छन्)
urls = [
    "https://www.onlinekhabar.com/2024/05/1476402",
    "https://ratopati.com/story/426955",
    "https://setopati.com/politics/330058"
]

scraped_data = []
for u in urls:
    print(f"Scraping: {u}")
    data = scrape_nepali_portal(u)
    if data:
        scraped_data.append(data)
    time.sleep(2) # ब्लक हुनबाट बच्न

# DataFrame मा देखाउने
df = pd.DataFrame(scraped_data)
df.head()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def get_onlinekhabar_links():
    """होमपेजबाट समाचारका लिङ्कहरू निकाल्ने फङ्सन"""
    url = "https://www.onlinekhabar.com/"
    headers = {'User-Agent': 'Mozilla/5.0'}

    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, 'html.parser')

    links = []
    # Onlinekhabar मा समाचारका लिङ्कहरू प्राय: <a> ट्याग भित्र हुन्छन्
    for a in soup.find_all('a', href=True):
        href = a['href']
        # केवल समाचारका लिङ्कहरू मात्र फिल्टर गर्ने (जसमा मिति वा समाचार ID हुन्छ)
        if "onlinekhabar.com/2026/" in href or "onlinekhabar.com/2025/" in href:
            if href not in links:
                links.append(href)

    return links

def scrape_article_details(url):
    """लिङ्क भित्र छिरेर टाइटल र टेक्स्ट निकाल्ने फङ्सन"""
    try:
        headers = {'User-Agent': 'Mozilla/5.0'}
        res = requests.get(url, headers=headers, timeout=10)
        soup = BeautifulSoup(res.content, 'html.parser')

        title = soup.find('h1').get_text(strip=True) if soup.find('h1') else None
        # समाचारको मुख्य भाग खोज्ने (ok18-single-post-content-wrap क्लास प्राय प्रयोग हुन्छ)
        paragraphs = soup.find_all('p')
        content = " ".join([p.get_text(strip=True) for p in paragraphs if len(p.get_text()) > 30])

        if title and len(content) > 100:
            return {"URL": url, "Title": title, "Text": content, "Label": "Real"}
    except:
        return None

# --- प्रक्रिया सुरु गरौं ---

print("⏳ होमपेजबाट लिङ्कहरू निकाल्दै...")
all_links = get_onlinekhabar_links()
print(f"✅ {len(all_links)} वटा लिङ्कहरू फेला परे।")

# परीक्षणको लागि पहिलो १० वटा लिङ्क मात्र स्क्र्याप गरौं (समय बचाउन)
# धेरै डेटा चाहिएमा [:10] हटाउनुहोस्
target_links = all_links[:10]

final_data = []
print("⏳ समाचार संकलन गर्दै (यसमा केही समय लाग्न सक्छ)...")

for link in target_links:
    data = scrape_article_details(link)
    if data:
        final_data.append(data)
        print(f"✔️ Scraped: {data['Title'][:50]}...")
    time.sleep(1) # सर्भर ब्लक हुनबाट बच्न १ सेकेन्डको ग्याप

# DataFrame मा रूपान्तरण
df_real = pd.DataFrame(final_data)

# नतिजा हेर्ने र सेभ गर्ने
print("\n🎉 स्क्र्यापिङ पुरा भयो!")
print(df_real.head())
df_real.to_csv('onlinekhabar_data.csv', index=False)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def scrape_nepal_fact_check():
    url = "https://nepalfactcheck.org/"
    headers = {'User-Agent': 'Mozilla/5.0'}

    try:
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')

        fake_news_list = []

        # नेपाल फ्याक्ट चेकमा समाचारका ब्लकहरू खोज्ने
        articles = soup.find_all('article')

        for article in articles:
            title_tag = article.find('h2') or article.find('h3')
            link_tag = article.find('a', href=True)

            if title_tag and link_tag:
                title = title_tag.get_text(strip=True)
                link = link_tag['href']

                # समाचार भित्र छिरेर विस्तृत विवरण तान्ने
                inner_res = requests.get(link, headers=headers)
                inner_soup = BeautifulSoup(inner_res.content, 'html.parser')

                # मुख्य दाबी (Claim) वा विवरण तान्ने
                paragraphs = inner_soup.find_all('p')
                content = " ".join([p.get_text(strip=True) for p in paragraphs if len(p.get_text()) > 20])

                fake_news_list.append({
                    "URL": link,
                    "Title": title,
                    "Text": content,
                    "Label": "Fake"
                })
                print(f"✔️ Fact-Checked Scraped: {title[:50]}...")
                time.sleep(1) # सर्भरलाई समय दिने

        return fake_news_list
    except Exception as e:
        print(f"Error: {e}")
        return []

# डाटा संकलन गरौं
print("⏳ नेपाल फ्याक्ट चेकबाट डाटा तान्दै...")
fake_data = scrape_nepal_fact_check()
df_fake = pd.DataFrame(fake_data)

print(f"\n✅ जम्मा {len(df_fake)} वटा 'Fake' न्युज संकलन भयो।")
```

```python
# दुवै डाटाफ्रेम जोड्ने (अघिल्लो स्टेपको df_real र अहिलेको df_fake)
# नोट: यदि df_real बनाउनु भएको छैन भने पहिले त्यो कोड रन गर्नुहोस्
combined_df = pd.concat([df_real, df_fake], ignore_index=True)

# डाटालाई तल-माथि पार्ने (Shuffle) ताकि मोडेलले राम्रोसँग सिक्न सकोस्
combined_df = combined_df.sample(frac=1).reset_index(drop=True)

# अन्तिम फाइल सेभ गर्ने
combined_df.to_csv('nepali_fake_news_master_dataset.csv', index=False)

print("🚀 मेसिन लर्निङका लागि मास्टर डाटासेट तयार भयो!")
print(combined_df.head())
```

```python
# केही सामान्य नेपाली स्टपवर्ड्सको सूची
nepali_stopwords = [
    'र', 'पनि', 'छ', 'छन्', 'हो', 'हुन्', 'भएको', 'गरेको', 'भने', 'गर्नु',
    'तथा', 'गरेका', 'भनेका', 'यस', 'यो', 'त्यो', 'मा', 'को', 'का', 'ले',
    'लाई', 'देखि', 'बाट', 'सम्म', 'भनेर', 'पर्ने', 'पर्छ', 'हुने', 'हुन्छ'
]
```

```python
import re

def clean_nepali_text(text):
    if not isinstance(text, str):
        return ""

    # १. विराम चिन्ह र विशेष क्यारेक्टरहरू हटाउने (। , ? ! आदि)
    # नेपाली युनिकोड रेन्ज: \u0900-\u097F
    text = re.sub(r'[^\u0900-\u097F\s]', '', text)

    # २. अङ्ग्रेजी शब्दहरू र अंकहरू हटाउने (यदि छन् भने)
    text = re.sub(r'[a-zA-Z0-9]', '', text)

    # ३. ह्वाइट स्पेस (अनावश्यक खाली ठाउँ) हटाउने
    text = " ".join(text.split())

    # ४. स्टपवर्ड्स हटाउने
    words = text.split()
    cleaned_words = [w for w in words if w not in nepali_stopwords]

    return " ".join(cleaned_words)

# मास्टर डाटासेटमा यो फङ्सन लागू गरौं
print("🧹 डाटा क्लीनिंग सुरु हुँदैछ...")
combined_df['Cleaned_Text'] = combined_df['Text'].apply(clean_nepali_text)

print("✅ क्लीनिंग पुरा भयो!")
print(combined_df[['Text', 'Cleaned_Text']].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF Vectorizer सेटअप
tfidf = TfidfVectorizer(max_features=5000) # धेरै प्रयोग हुने ५००० शब्द मात्र लिने

# टेक्स्टलाई म्याट्रिक्समा बदल्ने
X = tfidf.fit_transform(combined_df['Cleaned_Text']).toarray()
y = combined_df['Label']

print(f"डेटाको आकार (Shape): {X.shape}")
```

```python
from sklearn.model_selection import train_test_split

# X मा हाम्रो टेक्स्ट डाटा छ र y मा 'Real/Fake' लेबल छ
X_train, X_test, y_train, y_test = train_test_split(
    combined_df['Cleaned_Text'],
    combined_df['Label'],
    test_size=0.2,      # २०% डाटा जाँच्नका लागि राख्ने
    random_state=42     # हरेक पटक एउटै नतिजा आओस् भनेर
)

print(f"ट्रेनिङ डाटा संख्या: {len(X_train)}")
print(f"टेस्टिङ डाटा संख्या: {len(X_test)}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# पाइपलाइन बनाउने: पहिले शब्दलाई अंकमा बदल्ने (TF-IDF) र त्यसपछि मोडेल चलाउने
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('nb_classifier', MultinomialNB())
])

# मोडेललाई ट्रेनिङ दिने
model_pipeline.fit(X_train, y_train)

# टेस्ट डाटामा भविष्यवाणी गर्ने
predictions = model_pipeline.predict(X_test)

# नतिजा हेर्ने
print(f"मोडेलको एक्युरेसी (Accuracy): {accuracy_score(y_test, predictions)*100:.2f}%")
print("\nविस्तृत रिपोर्ट:\n", classification_report(y_test, predictions))
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# १. तालिमका लागि केही नेपाली समाचारका हरफहरू (हामीले अघि स्क्र्याप गरेको डेटा यहाँ राख्न सकिन्छ)
data = """नेपालमा भोलि सार्वजनिक बिदा दिने निर्णय भएको छ।
सरकारले नयाँ आर्थिक प्याकेज घोषणा गर्ने तयारी गरेको छ।
खेलकुद क्षेत्रमा नेपालले नयाँ सफलता हात पारेको छ।"""

# २. टोकनाइजेसन (शब्दलाई अंकमा बदल्ने)
tokenizer = Tokenizer()
tokenizer.fit_on_texts([data])
total_words = len(tokenizer.word_index) + 1

input_sequences = []
for line in data.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

# ३. प्याडिङ (सबै वाक्यलाई एउटै लम्बाइको बनाउने)
max_sequence_len = max([len(x) for x in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre'))

# ४. फिचर्स र लेबल छुट्याउने
X, y = input_sequences[:,:-1], input_sequences[:,-1]
y = tf.keras.utils.to_categorical(y, num_classes=total_words)

# ५. मोडेल बनाउने
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(total_words, 100, input_length=max_sequence_len-1),
    tf.keras.layers.LSTM(150),
    tf.keras.layers.Dense(total_words, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X, y, epochs=100, verbose=1)
```

```python
def generate_nepali_news(seed_text, next_words=5):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')
        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)

        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break
        seed_text += " " + output_word
    return seed_text

# प्रयोग गरेर हेरौं
print(generate_nepali_news("नेपालमा भोलि", 4))
```
